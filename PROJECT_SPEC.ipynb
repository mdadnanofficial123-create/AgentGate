# PROJECT SPECIFICATION: AgentGate (CI/CD AI Agent Guardrail)

> **Tagline:** The 15-Minute, Zero-Trust CI/CD Guardrail for AI Agents.

---

## 1. Project Overview & Vision

**AgentGate** is a lightweight, zero-trust evaluation gate for AI agents and LLM applications. It integrates directly into developer CI/CD workflows (e.g., GitHub Actions) to run automated, synthetic user conversations against staging AI agents before code or prompt updates reach production.

### Core Value Drivers
* **Zero-Trust Security:** Executes entirely inside the client's CI/CD runner—no customer data ever leaves their network.
* **Declarative Configuration:** YAML-first design allows Product Managers and QA leads to write test scenarios without writing code.
* **15-Minute Integration:** Drop a single `.github/workflows/agentgate.yml` action into any repository for instant coverage.

---

## 2. Problem Statement

AI agents are probabilistic. Updating a system prompt, modifying a RAG pipeline, or upgrading a model version often leads to unexpected side effects in production:

1. **Prompt Regression:** Fixing one edge case breaks handling for another.
2. **Multi-Turn Exploits:** Agents reveal private instructions after 3–5 back-and-forth messages.
3. **Silent Hallucinations & Schema Breaks:** Agents return invalid JSON or wrong facts.
4. **Manual Testing Bottleneck:** Engineers spend hours manually typing test messages into chat windows prior to release.

```text
┌─────────────────────────────────────────────────────────┐
│                    DEVELOPER REPOSITORY                 │
│                                                         │
│   Git Push / PR ──► GitHub Action ──► eval.yaml Specs   │
└───────────────────────────┬─────────────────────────────┘
                            │
                            ▼
┌─────────────────────────────────────────────────────────┐
│                    AGENTGATE RUNNER                     │
│                                                         │
│  ┌──────────────────┐        ┌───────────────────────┐  │
│  │  Simulated User  │ ─────► │  Staging Agent API    │  │
│  │ (Adversarial LLM)│ ◄───── │ (Target Application)  │  │
│  └──────────────────┘        └───────────────────────┘  │
└───────────────────────────┬─────────────────────────────┘
                            │
                            ▼
┌─────────────────────────────────────────────────────────┐
│                  EVALUATION & GATE PASS                 │
│                                                         │
│  1. Fast Regex / Schema Filters                         │
│  2. LLM-as-a-Judge Rubric (G-Eval Score)                │
│  3. HTML Dashboard + GitHub PR Status (Pass/Fail)       │
└─────────────────────────────────────────────────────────┘

---

## 4. Competitive Matrix

| Feature / Dimension | Braintrust / LangSmith | DeepEval / Promptfoo | AgentGate (Us) |
| :--- | :--- | :--- | :--- |
| **Data Privacy** | Cloud SaaS (Data leaves network) | Self-hosted options | **Zero-Trust (100% in local runner)** |
| **Setup Overhead** | Complex SDK integration | Writing Python test scripts | **15-Min YAML Configuration** |
| **Multi-Turn Attacks** | Manual / Complex scripting | Basic single-turn focus | **Native 5-Turn Synthetic Simulator** |
| **Pricing Model** | $250–$1,000+/mo SaaS | Freemium / Open source | **Flat License / One-Time Fee** |

---

## 5. Development Roadmap (Phased Plan)

### Phase 1: Core Engine & Local CLI (Week 1 / Days 1–5)
* **Goal:** A local CLI that parses an `eval.yaml` file, queries a target HTTP agent, runs basic regex + LLM-as-a-judge scoring, and logs pass/fail in the terminal.
* **Deliverable:** `npx agentgate run --config eval.yaml`

### Phase 2: Multi-Turn Adversarial Bot & HTML Report (Week 2 / Days 6–10)
* **Goal:** Add a 5-turn synthetic user simulator that attempts prompt injections, wrap everything in a GitHub Action, and generate a static HTML report.
* **Deliverable:** Working `.github/workflows/agentgate.yml` action.

### Phase 3: Pitch Package & Pre-Audit Demo (Week 3 / Days 11–14)
* **Goal:** Run the tool against the target company's public/staging agent endpoint to catch 2–3 real issues. Draft the 5-slide pitch deck and proposal.
* **Deliverable:** Live Pitch Meeting & Demo.

---

## 6. Daily Execution Plan (Week 1 Breakdown)

| Day | Goal / Task Focus | Deliverable |
| :--- | :--- | :--- |
| **Day 1** | Repository Setup, Types & YAML Parser | Local CLI loading `eval.yaml` |
| **Day 2** | Webhook / Agent HTTP Runner & Payload Dispatcher | Send inputs to target API & log latency/status |
| **Day 3** | Fast Deterministic Evaluator Engine | Regex, keyword blacklists, & JSON schema validators |
| **Day 4** | LLM-as-a-Judge Engine (Groq / Llama 3) | Rubric evaluation scoring (0.0 to 1.0) |
| **Day 5** | Terminal Output Formatting & CLI Polish | Clean color-coded Pass/Fail test logs |

---

┌─────────────────────────────────────────────────────────┐
│                    AGENTGATE TECH STACK                │
├───────────────────┬─────────────────────────────────────┤
│ Language & Runtime│ TypeScript / Node.js (v20+)         │
│ CLI Engine        │ Commander.js + Colorette (Color logs)│
│ Schema & Parsing  │ Zod (Validation) + yaml (Parser)    │
│ HTTP Client       │ Native fetch / Axios                │
│ LLM Judge API     │ Groq SDK / OpenAI SDK (Free Tier)   │
│ CI/CD Distribution│ GitHub Actions (Node 20 Runner)     │
└───────────────────┴─────────────────────────────────────┘


┌───────────────────────────────────────────────────────────┐
│                    DAY 1 EXECUTION PLAN                   │
├───────────────────────────────────────────────────────────┤
│ 1. Initialize TypeScript & Node.js environment            │
│ 2. Define the strict Zod configuration schema (`eval.yaml`)│
│ 3. Build the YAML file reader and parser                  │
│ 4. Build the Commander CLI entry point                    │
│ 5. Create a sample `eval.yaml` and verify local execution  │
└───────────────────────────────────────────────────────────┘


┌───────────────────────────────────────────────────────────┐
│                    DAY 2 EXECUTION PLAN                   │
├───────────────────────────────────────────────────────────┤
│ 1. Create the API Runner (HTTP Client)                    │
│ 2. Build assertion rules (Forbidden Terms & Latency)      │
│ 3. Execute test cases in parallel using `Promise.all`     │
│ 4. Output rich terminal results (Pass / Fail formatting)  │
│ 5. Wire the runner into the main `agentgate run` CLI      │
└───────────────────────────────────────────────────────────┘


┌─────────────────────────────────────────────────────────────┐
│                       DAY 3 PIPELINE                        │
├─────────────────────────────────────────────────────────────┤
│ 1. Parse LLM Evaluation Prompt from `eval.yaml`              │
│ 2. Execute target agent request (Existing Runner)           │
│ 3. Send (Input + Agent Response) to Judge LLM               │
│ 4. Parse Judge Verdict (JSON output: pass/fail + reasoning) │
│ 5. Aggregate with latency & regex rules in main CLI report  │
└─────────────────────────────────────────────────────────────┘


┌─────────────────────────────────────────────────────────────┐
│                      DAY 4 ROADMAP                          │
├─────────────────────────────────────────────────────────────┤
│ 1. Create `action.yml` (GitHub Action Metadata)             │
│ 2. Install `@actions/core` and `@actions/github`            │
│ 3. Build PR Commenter Engine (`src/reporter.ts`)            │
│ 4. Wire Action outputs into `src/index.ts`                  │
│ 5. Setup a sample GitHub Workflow (`.github/workflows/`)     │
└─────────────────────────────────────────────────────────────┘



┌─────────────────────────────────────────────────────────────┐
│                      DAY 5 ROADMAP                          │
├─────────────────────────────────────────────────────────────┤
│ 1. JSON Schema Assertion Validator (Zod Integration)        │
│ 2. Automated OWASP LLM Red-Teaming Engine                   │
│ 3. Updated Config Parser for `schema` and `security_suite`  │
│ 4. Verification with Real-World Test Cases                  │
└─────────────────────────────────────────────────────────────┘
